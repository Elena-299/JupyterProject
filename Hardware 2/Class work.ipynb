```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#EXPL: f(x) = sqrt(x) + 1

def f(x):
    return np.sqrt(x) + 1

x0 = 1
x1 = 3
width = x1 = x0
mid_point = (x0 + x1) / 2
height = f(mid_point)
area = width * height

print('Area:', area)
print(f"Analytical Solution: {(2/3)*(x1**1.5 - x0**1.5)+(x1-x0)}")

x_plot = np.linspace(0, 4, 100)
plt.plot(x_plot, f(x_plot), "red")
plt.fill([x0, x1, x1, x0], [0, 0, height, height], alpha=0.4)
plt.plot(mid_point, height, "bo")
plt.vlines(mid_point, 0, height, linestyle="--")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()
```

```python
def f(x):
    return 3*x
def g(x):
    return x**2
x = np.linspace(0,4,6)

AF = 0
AG = 0

for i in np.arange(len(x)-1):
    mid_point = (x[i] + x[i+1]) / 2
    AF = AF + (x[i+1] - x[i]* f(mid_point))
    AG = AG + (x[i+1] - x[i]* g(mid_point))

print(AF)
print(AG)

x_plot = np.linspace(0, 4, 100)
plt.plot(x_plot, f(x_plot), "blue")
for i in np.arange(len(x)-1):
    mid_point = (x[i] + x[i+1]) / 2
    plt.fill([x[i],x[i+1], x[i+1], x[i]], [0, 0, f(mid_point), f(mid_point)], alpha=0.4, color="red", edgecolor="black")

plt.grid()
plt.title("f(x) = 3x")
plt.show()
```

```python
#EXPL: f(x) = 3*x, g(x) = x**2
x = np.linspace(0, 4, 6) # 6 point nd 5 intervals

def f(x):
    return 3*x
y = f(x)

plt.plot(x, y, 'o-')
plt.vlines(x, 0, y)
plt.grid()
plt.xlabel("x")
plt.ylabel("y")
plt.show
```

```python
A = 0
for i in np.arange(len(x)-1):
    A = A + ((x[i+1] - x[i]) * (y[i+1] + y[i]) / 2)

print("numerical solution:", A)
print(f"Analytical solution: {(4 * 12)/2}")

plt.plot(x, y, "o-")
plt.vlines(x, 0, y)
plt.fill_between(x, y, color="red", alpha=0.4)
plt.grid()
plt.show()
```

```python
#EXPL: v(t) = 0.5 * t
ts = 0.1 # time step
t = np.arange(0, 5.01, ts)
v = 0.5 * t

plt.plot(t, v)
plt.grid()
plt.xlabel("time(t)")
plt.ylabel("velocity(m/s)")
plt.show()
```

```python
po = np.zeros(len(t)) #preserve an array of 0s with lenght of time
N = len(po)
for i in np.arange(N - 1):
    po[i+1] = po[i] + ts * (v[i+1] + v[i]) / 2

print("Numerical solution: ", po[-1])
print(f"Analytical solution: {0.25 *5**2}")

plt.plot(t, po)
plt.grid()
plt.xlabel("time")
plt.ylabel("velocity")
plt.show()
```
