### My solution

```python
# start your code here
x = np.arange(-10,10,0.001)
y1 = (x<-2.0) * 0
y2 = ((-2.0<=x) & (x<2.0)) * (2.0+x)
y3 = ((2.0<=x) & (x<6.0)) * (6.0-x)
y4 = (x>=6.0) * 0
y = y1 + y2 + y3 + y4

plt.plot(x,y, label= "Piecewise function", color= "r")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.legend()
plt.show
```

```python
# Your code starts here
x = np.arange(-10,10,0.001)
y1 = (x<-7.9)* (-383.02)
y2 = ((-7.9<=x) & (x<4.8)) * (1.0*x**3 + (-0.61)*x**2 + (-15.87*x) + 22.72)
y3 = (x>=4.8) * 43.08
y = y1 + y2 + y3

plt.plot(x,y, label= "piecewise function", color= "r")
plt.grid(True)
plt.ylabel("y")
plt.xlabel("x")
plt.legend()
plt.show()
```
