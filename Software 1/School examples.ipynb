```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,5,0.0005)
y = np.cos(x*np.pi)

fig,ax = plt.subplots(
    figsize = (5,3),
    constrained_layout = True)

ax.plot(x,y)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("sine wave")
fig.set_facecolor("white")
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,5,0.05)
y = np.cos(x*np.pi)

plt.figure(figsize = (3,2))
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Cos wave")
plt.xlim(0,10)
plt.ylim(-2,2)
plt.grid(True)
plt.show()


```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0,5,0.05)
y = np.sin(x*np.pi)

plt.axhline(0.7, color = "red") #horizontal H
plt.axvline(-0.7, color = "blue") #vertical V
plt.show()
#
#plt.savefig("sample.png")
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize = (7,4))
plt.plot(x, y1, label = "Sine Wave")
plt.plot(x, y2, label = "Cos Wave")

plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10,10,200)
y = x**3 - 2.0*x**2 -1.0*x + 2.0

plt.figure(figsize = (8,8))
plt.subplot(3,2,1)
plt.title("My first axes")
plt.plot(x,y)
plt.grid(True)

plt.subplot(3,2,2)
plt.plot(x,y)
plt.title("My limited axes")
plt.xlim(-4,4)
plt.ylim(-2,3)
plt.grid(True)
plt.show()
```
