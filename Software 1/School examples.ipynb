```python
# 4th order polynomial function
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,4)
y = 0.1 *x**4 - 0.5*x**3 + 0.2 *x**2 - 0.1*x + 4

fig, ax = plt.subplots(figsize = (10,7))

plt.plot(x,y, linestyle= "--", linewidth = 5, label = "4th order")
plt.grid(True)
plt.xlabel("X", fontsize = 15)
plt.ylabel("Y", fontsize = 15)
plt.title("Graph of function", fontsize = 24)
plt.legend()
plt.show()
```

```python
#multiple figures same code
import numpy as np
import matplotlib.pyplot as plt

fig1, ax1 = plt.subplots()
x = np.linspace(0,4)
y1 = x - 3

plt.plot(x,y1, color = "r", label = "x-3")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
#plt.show() if multiple figs have only one show at the end

y2 = x**2 - 4*x
fig2, ax2 = plt.subplots()
ax2.plot(x,y2, color = "b", label = "$x^2 - 4x$")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
#one fig multiple lines/axes
import numpy as np
import matplotlib.pyplot as plt

fig. ax = plt.subplots(2, figsize = (10,8))
x = np.linspace(0,4)
y1 = x - 3
plt.sca(ax[0])
plt.plot(x,y1, color = "r", label = "x-3")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
#plt.show() if multiple figs have only one show at the end


y2 = x**2 - 4*x
plt.sca(ax[1])
plt.plot(x,y2, color = "r", label = "x**2")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()

```

```python
#create graph using boolean values:
# y = 0.25*x**2
x = np.arange(-5,5)
y = 0.25*x**2
z = y>2

fig,ax = plt.subplots(2, figsize=(10,8))
#set current axis = sca
plt.sca(ax[0])
plt.plot(x,y, color="orange")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")

plt.sca(ax[1])
plt.plot(x,z, color="yellow")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```

```python
x = np.arange(-5,5,0.01)
y = 0.25*x**2
z = y>2
fig,ax = plt.subplots(3, figsize=(8,7), sharex=True, sharey=True)

plt.sca(ax[0])
plt.plot(x,y, color="orange")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")

plt.sca(ax[1])
plt.plot(x,z, color="yellow")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("z")

plt.sca(ax[2])
plt.plot(x,z, color="red")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("z")
```

```python
#piecewise functions
x = np.arange(-5,5,0.01)
y1 = x**2 - 15
y2 = x*3
#this is a boolean arithmetic aproach
y = (x<0)*y1 + (x>=0)*y2

fig, ax = plt.subplots(3, figsize= (10,9), sharex = True, sharey = True)

plt.sca(ax[0])
plt.plot(x,y1, label="y1")
plt.grid(True)
plt.legend(loc= "upper right")

plt.sca(ax[1])
plt.plot(x,y2, label="y2")
plt.grid(True)
plt.legend(loc= "upper right")

plt.sca(ax[2])
plt.plot(x,y, label="y")
plt.grid(True)
plt.legend(loc= "upper right")
```

```python
#explicit built-in piecewise function
x = np.linspace(0,10,100000)
y = np.piecewise(x,[(x<3) & (x<=6), x>=6], [0,1,0])
plt.plot(x,y)
plt.show()
```
