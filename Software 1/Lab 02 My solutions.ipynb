### My solution

```python
# Your code starts here
x = np.arange(0,10,0.01)
f = (2.3*x)**0.5
g = (x-4.19)**2
fig, ax = plt.subplots(3, figsize=(10,15))

plt.sca(ax[0])
plt.plot(x,f, label = "f(x)")
plt.plot(x,g, label = "g(x)")
plt.title("Both graph - overview")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.sca(ax[1])
plt.plot(x,f, label = "f(x)")
plt.plot(x,g, label = "g(x)")
plt.title("Intersection one - zoomed")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(2.62,2.63)
plt.ylim(2.45,2.47)
plt.legend()

plt.sca(ax[2])
plt.plot(x,f, label = "f(x)")
plt.plot(x,g, label = "g(x)")
plt.title("Intersection two - zoomed")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(6.12,6.13)
plt.ylim(3.75,3.76)
plt.legend()
plt.show()
```
