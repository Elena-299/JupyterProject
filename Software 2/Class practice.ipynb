```python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(6)
y = 10**x
print(y)

#exponential function with 2-base
y2 = 2**x
print(y2)

#exponential function with e-base
x3 = np.arange(8)
y3 = np.exp(x3)
y4 = np.exp(-x3)
print(y3)
print(y4)
```

```python
x = np.linspace(-2,2)
y = np.exp(-x)
y1 = np.exp(1.0*x)
y2 = np.exp(1.2*x)
y3 = np.exp(1.5*x)

plt.plot(x,y, label = "$y = e^(-x)$")
plt.plot(x,y1, label = "$y = e^(x)$")
plt.plot(x,y2, label = "$y = e^(1.2x)$")
plt.plot(x,y3, label = "$y = e^(1.5x)$")

plt.grid()
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
