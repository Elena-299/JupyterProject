Tips
- Here you **can use for-loops** for indexing both rows and columns of the array

- It is a good practise first to initialize the array. You can do that for example using the function `numpy.zeros()`.

```python
# My solution for problem 4
from numpy.linalg import det
A = np.zeros((9,9))
n = 0
m = 0
for r in range(9):
    n = n+1
    for c in range(9):
        m = m+1
        A[r,c] = (-1)**(n+m) * (n + (2*m))
print("A = \n", A)

determinant = det(A)
print("\nStep 2."
      "\nThe determinant of A is", determinant)
```

- Calculate the $det(A(x, a))$ and store the values to another variable, for example y. 
- Draw a graph of the determinant versus value of x, e.g. plot(x, y) 
- Finally, from the graph, find the value of x, when $y = 0$.
- **NOTE** $x = 0$ is NOT the answer we are looking for! 

The value for $a$ is given below for you.

```python
# Initialize random number generator
rng = np.random.default_rng(student_id+2)

# Generate random value for a
a = np.round(rng.uniform(3.4, 5.0), 2)
print(f'a = {a}')
```

```python
# Solution for problem 5
def matrix(a, x):
    return np.array([[a + x, a - x, a - x],[a - x, a + x, a - x],[a - x, a - x, a + x]])

def determinant(a, x):
    A = matrix(a, x)
    return np.linalg.det(A)

x = np.linspace(0, 20, 1000)
y = [determinant(a, i) for i in x]
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="determinant")
plt.axhline(0, color="red", linestyle="--")
plt.xlabel('x')
plt.ylabel('Determinant')
plt.title("Determinant and x graph")
plt.legend()
plt.grid(True)
plt.xlim(14.3, 14.5)
plt.ylim(-50,50)
plt.show()

print(f"When a is equal to {a}, and the determinant is 0, x is 14.4")
```

- a and x are related so that for any value of a, x is going to be exactly 3 times bigger
